```python
import numpy as np
import control as ctrl
import matplotlib.pyplot as plt
```

```python
# Define Transfer Function coefficients
# G(s) = 1 / (s^4 + 2s^3 + 3s^2 + 2s)
num = [1]
den = [1, 2, 3, 2, 0]
```

```python
# Create the system transfer function
system = ctrl.TransferFunction(num, den)
```

```python
# Extract Poles and Zeros
poles = system.poles()
zeros = system.zeros()
```

```python
# Print System Properties
print("Transfer Function G(s):")
print(system)
print("\nPoles of the system:")
for p in poles:
    print(np.round(p, 3))
print("\nZeros of the system:")
print(zeros if len(zeros) > 0 else "None")
```

```python
# Plot 1: Step Response
plt.figure(figsize=(12, 8))
t, y = ctrl.step_response(system)
plt.plot(t, y, label='Step Response')
plt.title('Step Response of the System')
plt.xlabel('Time (s)')
plt.ylabel('Output')
plt.grid(True)
plt.legend()
plt.show()
```

```python
# Plot 2: Pole-Zero Map
plt.figure(figsize=(12, 8))
# Plot zeros as 'O' and poles as 'X'
if len(zeros) > 0:
    plt.scatter(np.real(zeros), np.imag(zeros), s=100, marker='o', color='blue', label='Zeros')
plt.scatter(np.real(poles), np.imag(poles), s=100, marker='x', color='red', label='Poles')
```
