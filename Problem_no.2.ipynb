```python
# Import necessary libraries for control theory and visualization
import control as ctrl
import matplotlib.pyplot as plt
import numpy as np
```

```python
# Define the transfer function coefficients
# G(s) = (s + 1) / (s^4 + 4s^3 + 7s^2 + 6s + 3)
num = [1, 1]
den = [1, 4, 7, 6, 3]
```

```python
# Create the transfer function object
sys = ctrl.TransferFunction(num, den)
```

```python
# 1. Print system details
print("--- Control System Analysis ---")
print("Transfer Function G(s):")
print(sys)
print(f"Zeros: {ctrl.zeros(sys)}")
print(f"Poles: {ctrl.poles(sys)}")
```

```python
# 2. Setup plotting
plt.figure(figsize=(14, 10))
```

```python
# Pole-Zero Map
plt.subplot(2, 2, 1)
ctrl.pzmap(sys, plot=True)
plt.title('Pole-Zero Map')
plt.grid(True)
```

```python
# Step Response
plt.subplot(2, 2, 2)
t, y = ctrl.step_response(sys)
plt.plot(t, y, 'b', linewidth=2)
plt.title('Step Response')
plt.xlabel('Time (sec)')
plt.ylabel('Output x2(t)')
plt.grid(True)
```

```python
# Bode Plot - Magnitude
plt.subplot(2, 2, 3)
mag, phase, omega = ctrl.bode_plot(sys, plot=False)
plt.semilogx(omega, 20 * np.log10(mag), 'r', linewidth=2)
plt.title('Bode Plot (Magnitude)')
plt.xlabel('Frequency (rad/sec)')
plt.ylabel('Magnitude (dB)')
plt.grid(True)
```

```python
# Bode Plot - Phase
plt.subplot(2, 2, 4)
plt.semilogx(omega, np.rad2deg(phase), 'g', linewidth=2)
plt.title('Bode Plot (Phase)')
plt.xlabel('Frequency (rad/sec)')
plt.ylabel('Phase (deg)')
plt.grid(True)
```

```python
plt.tight_layout()
plt.show()
```
