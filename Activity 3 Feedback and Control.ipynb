```python
import control as ctrl
import matplotlib.pyplot as plt

num = [5, 5]        # represents 5s + 5
den = [1, 4, 3]     # represents s^2 + 4s + 3

G = ctrl.tf(num, den)
print(G)
```

```python
poles = ctrl.poles(G)
print("Poles:", poles)
```

```python
zeros = ctrl.zeros(G)
print("Zeros:", zeros)
```

```python
import matplotlib.pyplot as plt

ctrl.pzmap(G, grid=True)
plt.show()
```
