```python
from openai import OpenAI
from langchain_community.chat_message_histories import ChatMessageHistory

# 1. Connect to LM Studio local server
client = OpenAI(
    base_url="http://localhost:1234/v1",
    api_key="lm-studio"  
)
```

```python
history = ChatMessageHistory()

print("Chat started! Look at the TOP CENTER of VS Code for the input box. Type 'exit' to stop.\n")

while True:
    
    query = input("User (Type 'exit' to quit): ")

    # Check for 'exit' command
    if query.lower() == 'exit':
        print("\nExiting...")
        print("Program Finished.")
        break
        
    print("User Input: " + query)
    print('- ' * 50)
    
    if not query.strip():
        continue
        
    # Add user message to LangChain history
    history.add_user_message(query)

    # Convert LangChain 'human/ai' roles to OpenAI 'user/assistant' roles
    messages_with_history = []
    for m in history.messages:
        role = "user" if m.type == "human" else "assistant"
        messages_with_history.append({"role": role, "content": m.content})

    # Send the request to LM Studio
    stream = client.chat.completions.create(
        model="google/gemma-4-e2b",  # Matches your LM studio model
        messages=messages_with_history,
        stream=True
    )
    
    full_ai_response = ""       

    # Print the response chunk by chunk as it generates
    for chunk in stream:
        if chunk.choices[0].delta.content:
            content = chunk.choices[0].delta.content
            print(content, end="", flush=True)
            full_ai_response += content     

    # Save the AI response back to LangChain history
    history.add_ai_message(full_ai_response)
    print('\n ')     
    print('=' * 100)
    print(' ')
```
