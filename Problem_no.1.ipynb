```python
import numpy as np
import control as ctrl
import matplotlib.pyplot as plt
```

```python
# Define the numerator and denominator coefficients
num = [1, 1]
den = [1, 4, 7, 6, 3]
```

```python
# Create the transfer function
sys = ctrl.TransferFunction(num, den)
```

```python
# Extract Poles and Zeros
poles = ctrl.poles(sys)
zeros = ctrl.zeros(sys)
```

```python

# Print system properties
print("Transfer Function G(s):")
print(sys)
print("\nPoles:")
for p in poles:
    print(f"  {p:.4f}")
print("\nZeros:")
for z in zeros:
    print(f"  {z:.4f}")
```

```python
# Plot 1: Step Response
plt.figure(figsize=(12, 8))
t, y = ctrl.step_response(sys)
plt.plot(t, y)
plt.title('Step Response of the System')
plt.xlabel('Time (s)')
plt.ylabel('Output')
plt.grid(True)
plt.show()
```

```python
# Plot 2: Pole-Zero Map
plt.figure(figsize=(12, 8))
ctrl.pzmap(sys, plot=True)
plt.title('Pole-Zero Map')
plt.xlabel('Real')
plt.ylabel('Imaginary')
plt.grid(True)
```
