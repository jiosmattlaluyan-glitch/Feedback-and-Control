```python
import numpy as np
import control as ctrl
import matplotlib.pyplot as plt
```

```python
# Define the transfer function coefficients
# G(s) = 3 / (8s^4 + 12s^3 + 26s^2 + 18s)
num = [3]
den = [8, 12, 26, 18, 0]
```

```python
# Create the transfer function object
sys = ctrl.TransferFunction(num, den)
```

```python
# Calculate poles and zeros
poles = ctrl.poles(sys)
zeros = ctrl.zeros(sys)
```

```python
# Print system properties
print("Transfer Function G(s):")
print(sys)
print("\nPoles:")
for p in poles:
    print(f"  {p}")
print("\nZeros:")
if len(zeros) == 0:
    print("  None")
else:
    for z in zeros:
        print(f"  {z}")
```

```python
# Generate Step Response Plot
plt.figure(figsize=(12, 8))
t, y = ctrl.step_response(sys)
plt.plot(t, y, label='Step Response', color='blue', linewidth=2)
plt.title('Step Response of the System')
plt.xlabel('Time (s)')
plt.ylabel('Output')
plt.grid(True)
plt.legend()
plt.show()
```

```python
# Generate Pole-Zero Map Plot
plt.figure(figsize=(12, 8))
ctrl.pzmap(sys, plot=True, title='Pole-Zero Map')
plt.title('Pole-Zero Map')
plt.xlabel('Real')
plt.ylabel('Imaginary')
plt.grid(True)
plt.show()
```
